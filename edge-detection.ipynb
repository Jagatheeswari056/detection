```python
import cv2 #Imports OpenCV for edge detection
import matplotlib.pyplot as plt #Used to display images

image_paths = [
    '/kaggle/input/edge-detector/building.jpg',
    '/kaggle/input/edge-detector/rose.jpg',
    '/kaggle/input/edge-detector/roses.jpg'
]

for path in image_paths: #This loop runs once per image
    img = cv2.imread(path) #Loads the image from disk

    if img is None:
        print(f"❌ Could not load {path}")
        continue

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) #Converts color image to gray image
    #100 → lower threshold,200 → upper threshold
    edges = cv2.Canny(gray, 100, 200) #Detects edges (boundaries) in the image

    plt.imshow(edges, cmap='gray') #Displays the detected edges
    #path.split("/")[-1] → extracts filename only
    plt.title(f'Edge Detection - {path.split("/")[-1]}') 
    plt.axis('off') #Hides x/y axis numbers
    plt.show() #Renders the image
```
